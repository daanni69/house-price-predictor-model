```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load the Dataset
path = "California_House_Prices_Data.csv"
df = pd.read_csv(path)
df.head()
```

```python
#Checking the info of the dataset
df.info()
```

```python
# Quick statistics
df.describe()
```

```python

# Missing values calculate
missing_values = df.isnull().sum().reset_index()
missing_values.columns = ["column", "missing_count"]
missing_values = missing_values[missing_values["missing_count"] > 0]

if not missing_values.empty:
    fig = px.bar(
        missing_values,
        x="column",
        y="missing_count",
        text="missing_count",
        title="Missing Values per Column",
        color="missing_count",
        color_continuous_scale="Reds"
    )
    fig.update_traces(textposition="outside")

    fig.update_layout(width=700, height=500)

    fig.show()
else:
    print("No missing values found!")
```

```python
#Checking for duplicates
duplicate_count = df.duplicated().sum()
if duplicate_count > 0:
    print(f"Number of duplicate rows: {duplicate_count}")
else:
    print("No duplicate rows found!")
```

```python
# Distribution of Target Variable
fig = px.histogram(
    df,
    x="median_house_value",
    nbins=50,
    marginal="box",     # boxplot on top
    opacity=0.7,
    title="🏠 Distribution of Median House Value",
    color_discrete_sequence=["#1f77b4"]  # single elegant blue shade
)

# Add median line (dark green instead of red)
median_value = df["median_house_value"].median()
fig.add_vline(
    x=median_value,
    line_width=2, line_dash="dash", line_color="darkgreen",
    annotation_text=f"Median: {median_value:,.0f}", annotation_position="top right"
)

# Layout updates
fig.update_layout(
    template="plotly_white",
    height=500, width=800,
    title_x=0.5,
    xaxis_title="House Value ($)",
    yaxis_title="Frequency",
    font=dict(size=14),
    margin=dict(l=40, r=40, t=60, b=40)
)

fig.show()
```

```python
import plotly.figure_factory as ff

# Population distribution
fig = ff.create_distplot(
    [df['population']], 
    group_labels=['Population'], 
    colors=['#00BFFF'], # Professional blue shade
    bin_size=200
)

# Layout customize
fig.update_layout(
    title="Population Distribution",
    title_x=0.5,
    template="plotly_white",
    xaxis_title="Population",
    yaxis_title="Density",
    plot_bgcolor="white",
    paper_bgcolor="white",
    font=dict(size=14),
    margin=dict(l=50, r=50, t=80, b=50),
)

fig.show()
```

```python
# Value counts for ocean_proximity
cat_counts = df["ocean_proximity"].value_counts().reset_index()
cat_counts.columns = ["ocean_proximity", "count"]

# Sort by count
cat_counts = cat_counts.sort_values(by="count", ascending=False)

# Plotly bar chart (Dashboard Style)
fig = px.bar(
    cat_counts,
    x="ocean_proximity",
    y="count",
    title="🌊 Ocean Proximity Distribution",
    text="count",   # show counts on bars
    color="ocean_proximity",
    color_discrete_sequence=px.colors.qualitative.Set2  # clean color set
)

# Layout updates
fig.update_traces(textposition="outside")  # counts above bars
fig.update_layout(
    template="plotly_white",
    showlegend=False,        # hide duplicate legend
    height=500, width=800,   # normal size
    title_x=0.5,             # center title
    margin=dict(l=40, r=40, t=60, b=40),
    font=dict(size=14)
)

fig.show()
```

```python
# Ocean Proximity vs House Value
fig3 = go.Figure()
for cat in df["ocean_proximity"].unique():
    fig3.add_trace(
        go.Box(
            y=df.loc[df["ocean_proximity"] == cat, "median_house_value"],
            name=cat, boxmean=True
        )
    )
fig3.update_layout(title="House Value by Ocean Proximity", width=700, height=500)
fig3.show()
```

```python
# Population vs House Value
fig2 = go.Figure(
    data=go.Scatter(
        x=df["population"],
        y=df["median_house_value"],
        mode="markers",
        marker=dict(
            size=df["households"]**0.5,
            color=df["median_income"],
            colorscale="Plasma",
            showscale=True,
            colorbar=dict(title="Median Income")
        ),
        opacity=0.6
    )
)
fig2.update_layout(title="Population vs House Value", width=700, height=500)
fig2.show()
```

```python
import plotly.figure_factory as ff

# Data for KDE plot
data = [df["housing_median_age"].dropna()]

# Create distribution plot (Histogram + KDE)
fig = ff.create_distplot(
    data, group_labels=["Housing Median Age"], bin_size=1, show_hist=True, show_rug=False
)

# Update layout for dashboard style
fig.update_layout(
    title="Distribution of Housing Median Age with KDE",
    template="plotly_white",   # clean dashboard style
    bargap=0.05,
    height=500, width=800,     # normal size
    xaxis_title="Housing Median Age",
    yaxis_title="Density",
    font=dict(size=14),
    title_x=0.5,               # center title
)

fig.show()
```

```python
# Geo-visualization: Houses on California map
fig = px.scatter_mapbox(
    df,
    lat="latitude",
    lon="longitude",
    size="population",
    color="median_house_value",
    hover_name="ocean_proximity",
    hover_data={
        "housing_median_age": True,
        "total_rooms": True,
        "households": True,
        "latitude": False,   # latitude/lon hide in hover
        "longitude": False
    },
    color_continuous_scale=px.colors.sequential.Plasma,
    size_max=15,
    zoom=5,
    height=600,
    title="🏠 Housing Prices in California"
)

# Dashboard style map
fig.update_layout(
    mapbox_style="carto-positron",  # clean background style
    margin=dict(l=20, r=20, t=60, b=20),
    title_x=0.5,   # center title
    font=dict(size=14),
    coloraxis_colorbar=dict(
        title="Median House Value",
        tickprefix="$"
    )
)

fig.show()
```

```python
# Scatter: Income vs House Value
fig = px.scatter(
    df, 
    x="median_income", 
    y="median_house_value",
    color="ocean_proximity",   # ab original column use karenge
    title="🏡 House Value vs Median Income",
    opacity=0.6,
    size_max=6,
    color_discrete_sequence=["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd"]  
    # aap yahan apne pasand ke colors dal sakte ho (hex ya naam)
)

# Layout updates
fig.update_layout(
    template="plotly_white",
    height=500, width=800,
    title_x=0.5,
    xaxis_title="Median Income",
    yaxis_title="Median House Value ($)",
    font=dict(size=14),
    margin=dict(l=40, r=40, t=60, b=40),
    legend_title="Ocean Proximity"
)

fig.show()
```

```python
# Histogram: Median Income
fig4 = go.Figure(
    data=go.Histogram(
        x=df["median_income"],
        nbinsx=40,
        marker=dict(color="#00CC96", line=dict(color="black", width=1))
    )
)
fig4.update_layout(title="Distribution of Median Income", width=700, height=500)
fig4.show()
```

```python
# Numerical columns list
num_cols = ['longitude', 'latitude', 'housing_median_age', 
            'total_rooms', 'total_bedrooms', 'population', 
            'households', 'median_income', 'median_house_value']

# Subplot grid (3 rows, 3 cols)
fig = make_subplots(
    rows=3, cols=3,
    subplot_titles=num_cols,
    horizontal_spacing=0.12,
    vertical_spacing=0.15
)

# Add boxplots for each feature
row, col = 1, 1
for i, feature in enumerate(num_cols):
    fig.add_trace(
        go.Box(y=df[feature], name=feature, boxpoints="outliers"),
        row=row, col=col
    )
    col += 1
    if col > 3:  # move to next row after 3 cols
        row += 1
        col = 1

# Layout settings
fig.update_layout(
    title="Boxplots of All Numerical Features (Outlier Detection)",
    title_x=0.5,
    template="plotly_white",
    height=900,
    width=1200,
    showlegend=False,
    font=dict(size=12),
)

fig.show()
```

```python
# Correlation Heatmap
corr = df[['housing_median_age','total_rooms','total_bedrooms',
           'population','longitude', 'latitude', 'households','median_income','median_house_value']].corr()

fig1 = go.Figure(
    data=go.Heatmap(
        z=corr.values,
        x=corr.columns,
        y=corr.columns,
        colorscale="Viridis",
        text=corr.values.round(2),
        texttemplate="%{text}",
        colorbar=dict(title="Correlation")
    )
)
fig1.update_layout(title="Correlation Heatmap", width=700, height=500)
fig1.show()
```
